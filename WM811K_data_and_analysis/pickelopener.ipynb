```python
import numpy as np

data = np.load("wm811k_wafer_maps.npz", allow_pickle=True)
wafer_maps = data["wafer_maps"]
```

```python
# visualize a random wafer and show its numerical data

import numpy as np
import matplotlib.pyplot as plt

# randomly choose a wafer index every time the cell runs
wafernum = np.random.randint(0, len(wafer_maps))

# convert wafer number to string for plot title
wafername = str(wafernum)

# get the wafer matrix from the dataset
wafer = wafer_maps[wafernum]

# print dataset explanation
print("Die value meanings:")
print("0 = background (outside wafer)")
print("1 = normal functioning die")
print("2 = defective die\n")

# print the raw wafer matrix so we can see the numbers
print("Numerical wafer data:")
print(wafer)

# create a figure to display the wafer image
plt.figure(figsize=(4,4))

# convert the matrix into an image
plt.imshow(wafer, cmap="gray")

# show color scale for die values
plt.colorbar(label="Die value")

# show wafer index number in title
plt.title("Wafer: " + wafername)

# hide axis numbers to make it cleaner
plt.axis("off")

# display the plot
plt.show()
```
